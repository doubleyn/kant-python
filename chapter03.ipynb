````python
# 리스트 : 가변 데이터 저장
shopping_list = ["apple", "banana", "cherry"]

print(shopping_list)
print(shopping_list[0])

shopping_list[0] = "kiwi"
print(shopping_list[0])
print(shopping_list)
````

### 리스트 메서드

````python
fruits = ["apple"]

# append
fruits.append("banana")
print(fruits)

# insert
fruits.insert(0, "kiwi")
print(fruits)

# remove
fruits.remove("apple")
print(fruits)

# fruits.remove("grape") # ValueError: list.remove(x): x not in list
````

### 튜플 (Tuple)

````python
axis = (38.123, 41.123)

# axis[0] = 36.1234 # TypeError: 'tuple' object does not support item assignment

point = 1, 2, 3 # 괄호가 없어도 튜플로 인식
print(point, type(point))
````

````python
import sys
print(sys.getsizeof([])) # 56
print(sys.getsizeof([1, 2, 3])) # 88
print(sys.getsizeof((1, 2, 3))) # 72 - 튜플 : 데이터 안정성 목적. 리스트보다는 튜플의 사이즈가 가볍다.
````

### 슬라이싱

````python
numbers = [0, 1, 2, 3, 4, 5]

# 1번 부터 4번 미만까지
print(numbers[1:4]) # [1, 2, 3] 마지막 인덱스는 exclusive

# 2칸 간격으로 추출
print(numbers[::2]) # [0, 2, 4]

# 역순으로 추출
print(numbers[::-1]) # [5, 4, 3, 2, 1, 0]

# 앞에서 3개만 추출
print(numbers[:3]) # [0, 1, 2]

# 3번부터 끝까지
print(numbers[3:]) # [3, 4, 5]

# 뒤에서 3개만 추출
print(numbers[-3:]) # [3, 4, 5]
````
