### 변수 이름 규칙

- 숫자로 시작 불가: 1_name (❌)
- 공백 불가: user name (❌)
- 하이픈(-) 불가: user-age (❌)
- 예약어 사용불가: if, def (❌)


### 컨벤션
- 스네이크 표기법: user_name, value_list

### 다른 표기법
- 파스칼 케이스: UserName, ValueList
- 카멜 케이스 (단봉낙타법): userName, valueList

````python
# 정수 자료형
age = 25
# 실수 자료형
pi = 3.14
# 불리언 자료형(Boolean)
is_alive = True
# 문자 자료형
user_name = "YoonSooYoung"
# 배열 Array
grade_type = [1, 2, 3, 4]
````
